#**4. The unit of analysis, as a real dataframe**

The unit of analysis is one individual content item (webpage/article). Each row represents one pseudonymized content page belonging to a client. The dataset contains information about content age, last update time, search impressions, clicks, sessions, click-through rate, and average search position.

I created a declining_proxy column to represent whether the content item is currently classified as declining. The content_id and client_id columns are identifiers used for grouping and tracking records, not model features.

```python
import pandas as pd

# Load the starter dataset
df = pd.read_csv("https://raw.githubusercontent.com/MubashshiraKaisar/FlyRank-AI-Tasks/main/data/raw/content_refresh_anonymized.csv")

# Create our target proxy
df["declining_proxy"] = (
    df["trend_direction"] == "down"
).astype(int)

# Check dataset dimensions
print("Dataset shape:", df.shape)

# Display sample content items
display(
    df[
        [
            "content_id",
            "client_id",
            "content_type",
            "content_age_days",
            "days_since_last_update",
            "impressions_90d",
            "clicks_90d",
            "sessions_90d",
            "ctr",
            "avg_position",
            "declining_proxy"
        ]
    ].head(10)
)
```

#**5. Why ML beats a fixed rule here**

A fixed rule might prioritize every content page that has not been updated for more than 180 days. However, content performance depends on several factors, including search impressions, engagement, content age, competition, and previous update history.

A page may be old but still perform well, while a recently updated page may already be losing traffic. Machine learning can identify patterns across multiple signals and their combinations instead of relying on one manually selected threshold.

This could help editors make more informed prioritization decisions. However, I would compare the ML approach against a simple rule-based baseline before claiming that ML performs better. If the fixed rule performs equally well, using ML may not be necessary.
