#**3. Initial Dataset Exploration**

The starter dataset contains **30,000 content pages and 44 columns**, covering search performance, content characteristics, and user engagement metrics.

After applying the eligibility criteria of having more than zero impressions and a content age of at least 90 days, all 30,000 pages remained eligible. Among these pages, **16,262 (54.21%) showed a declining performance trend**, while the average number of impressions over the last 90 days was **5,200.37 per page**.

These initial observations show that the dataset contains a substantial number of pages with declining trends. This makes it useful for investigating which observable signals are associated with declining performance and developing a system to prioritize pages for content review.

```python
import pandas as pd

data_path = "https://raw.githubusercontent.com/MubashshiraKaisar/FlyRank-AI-Tasks/main/data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(data_path)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
print(df.head())
```

```python
# Basic dataset information
total_rows = len(df)
total_pages = df["content_id"].nunique()

# Apply the starter dataset eligibility criteria
eligible_df = df[
    (df["impressions_90d"] > 0) &
    (df["content_age_days"] >= 90)
].drop_duplicates(subset="content_id")

eligible_pages = len(eligible_df)

# Calculate declining pages
declining_pages = (
    eligible_df["trend_direction"] == "down"
).sum()

declining_percentage = (
    declining_pages / eligible_pages * 100
    if eligible_pages > 0 else 0
)

# Calculate average impressions
average_impressions = eligible_df["impressions_90d"].mean()

print("===== STARTER DATASET OVERVIEW =====")
print(f"Total dataset rows: {total_rows:,}")
print(f"Unique content pages: {total_pages:,}")
print(f"Eligible content pages: {eligible_pages:,}")

print("\n===== KEY PERFORMANCE MEASUREMENTS =====")
print(f"Pages with declining trends: {declining_pages:,}")
print(f"Declining pages percentage: {declining_percentage:.2f}%")
print(f"Average 90-day impressions: {average_impressions:,.2f}")
```
